```python
#Criando uma base com as duas informações juntas
dados = pd.merge(paymants, orders, on='order_id', how='inner')
dados.head()
```

```python
#Filtrando somente os campos necessarios
colunas = ['order_id', 'order_status', 'order_purchase_timestamp',
           'payment_type', 'payment_value', 'payment_sequential']
dados = dados[colunas]
dados.info()
```

```python
# Removendo erros da base
dados = dados[dados['payment_type'] != 'not_defined']
dados = dados[dados['payment_value'] > 0]
dados.info()
```

```python
mapa_status = {
    'delivered': 'concluida',
    'canceled': 'cancelada',
    'unavailable': 'cancelada',
    'shipped': 'pendente',
    'invoiced': 'pendente',
    'processing': 'pendente',
    'approved': 'pendente',
    'created': 'pendente',
}

dados['status'] = dados['order_status'].map(mapa_status)
dados['status'].value_counts(dropna=False)
```

```python
print(dados[dados['status'].isnull()]['order_status'].unique())
```

```python
dados = dados.drop(columns=['order_status'])
dados.head()
```

```python
dados['order_purchase_timestamp'] = pd.to_datetime(dados['order_purchase_timestamp'])
dados.info()
```

```python
dados['mes'] = dados['order_purchase_timestamp'].dt.to_period('M')
dados.groupby('mes')['order_id'].nunique()
```

```python
data = dados['order_purchase_timestamp']
periodo = dados[(data >='2018-01-01') & (data < '2018-07-01')]

periodo.drop_duplicates('order_id')['status'].value_counts()
```

```python
pedidos = periodo[['order_id', 'status']].drop_duplicates()
quantidades = {'concluida':400, 'cancelada': 25, 'pendente':20}

sorteados = []
for status, n in quantidades.items():
    grupo = pedidos[pedidos['status'] == status]
    n = min(n, len(grupo))
    sorteados.append(grupo.sample(n=n, random_state=42))

ids_sorteados = pd.concat(sorteados)['order_id']

amostra =  periodo[periodo['order_id'].isin(ids_sorteados)]

amostra.drop_duplicates('order_id')['status'].value_counts()
```

```python
amostra = amostra.copy()
amostra['total_partes'] = amostra.groupby('order_id')['order_id'].transform('size')
```

```python
amostra.head()
```

```python
mapa_pagamento = {
    'credit_card': 'Cartão de crédito',
    'debit_card': 'Cartão de débito',
    'boleto': 'Boleto',
    'voucher': 'Voucher',
}
```

```python
descricao = 'Pedido' + amostra['order_id'].str[:8]

varias_partes = amostra["total_partes"] > 1
descricao[varias_partes]=(
    descricao[varias_partes] + "( pagamento "+ amostra["payment_sequential"].astype(str) + " de "+ amostra['total_partes'].astype(str) +")"
)
```

```python
receitas = pd.DataFrame({
    'id': 'rec_' + amostra['order_id'] + "_" + amostra['payment_sequential'].astype(str),
    'tipo': 'receita',
    'valor': amostra['payment_value'].round(2),
    'data': amostra['order_purchase_timestamp'].dt.strftime('%Y-%m-%dT%H:%M:%S'),
    'categoria':'Vendas',
    'formaPagamento': amostra['payment_type'].map(mapa_pagamento),
    'status': amostra['status'],
    'descricao': descricao,
})
```

```python
print(receitas.head())
print(receitas['id'].is_unique)
print(receitas['formaPagamento'].isnull().sum())
print(receitas['status'].value_counts())
```

```python
#Calcular vendas do mes

import numpy as np

rng = np.random.default_rng(42)

concluidas = amostra[amostra['status'] == 'concluida']
vendas_mes = concluidas.groupby('mes')['payment_value'].sum()
credito_mes = concluidas[concluidas['payment_type'] == 'credit_card'].groupby('mes')['payment_value'].sum()
debito_mes = concluidas[concluidas['payment_type'] == 'debit_card'].groupby('mes')['payment_value'].sum()

media = vendas_mes.mean()
print(vendas_mes)
print(f'Média mensal: R$ {media:.2f}')
```

```python
#Despesas
fixas = [
    # (chave, categoria, % da média, dia do mês, forma de pagamento, variação)
    ('aluguel',  'Aluguel',  0.15,  10, 'Boleto',        0.00),
    ('salarios', 'Salários', 0.25,   5, 'Transferência', 0.00),
    ('energia',  'Energia',  0.04,  15, 'Boleto',        0.10),
    ('internet', 'Internet', 0.015, 20, 'Boleto',        0.00),
]
```

```python
def nova_despesa(mes, chave, categoria, valor, dia, forma):
    data = mes.to_timestamp() + pd.Timedelta(days= dia -1)
    return {
        'id': f'desp_{mes}_{chave}',
        'tipo': 'despesa',
        'valor': round(valor, 2),
        'data': data.strftime('%Y-%m-%dT%H:%M:%S'),
        'categoria': categoria,
        'formaPagamento': forma,
        'status': 'concluida',
        'descricao': f'{categoria} - {mes.strftime("%m/%Y")}',
    }
```

```python
lista = []
for mes, venda in vendas_mes.items():
    for chave, categoria, pct, dia, forma, variacao in fixas:
        valor = media * pct * rng.uniform(1 - variacao, 1 + variacao)
        lista.append(nova_despesa(mes, chave, categoria, valor, dia, forma))
    valor_fornecedores = venda * rng.uniform(0.32, 0.38)
    lista.append(nova_despesa(mes, 'forncedores', 'Fornecedores', valor_fornecedores, 25, 'Boleto'))

    valor_taxas = credito_mes.get(mes, 0) * 0.03 + debito_mes.get(mes, 0) * 0.015
    lista.append(nova_despesa(mes,'taxa', 'Taxas de cartão', valor_taxas, 28, 'Debito automático'))

despesas = pd.DataFrame(lista)
despesas.head()
```

```python
despesas_mes = despesas.groupby(pd.to_datetime(despesas['data']).dt.to_period('M'))['valor'].sum()

resumo = pd.DataFrame({'receitas': vendas_mes, 'despesas': despesas_mes})
resumo['lucro'] = resumo['receitas'] - resumo['despesas']
resumo['margem_%'] = (resumo['lucro']/resumo['receitas'] * 100).round(1)

despesas.head()
print(resumo)
```

```python
transacoes = pd.concat([receitas, despesas], ignore_index=True)
transacoes = transacoes.sort_values('data').reset_index(drop=True)
```

```python
assert transacoes.isnull().sum().sum()==0
assert transacoes['id'].is_unique
assert transacoes['tipo'].isin(['receita', 'despesa']).all()
assert transacoes['status'].isin(['concluida', 'cancelada', 'pendente']).all()
assert (transacoes['valor']> 0).all()

print(transacoes.groupby([ 'tipo', 'status']).size())
```
